```python
#Import libraries
import pandas as pd
import numpy as np
import sklearn as sk
```

```python
#Import data
data = pd.read_csv('data_transactions.csv')
data.head(5)
```

```python
#Total data rows
print("Row count is:", data.shape[0])
```

```python
#View data columns
data.columns
```

```python
#Define X and y
X = data[['step', 'type', 'amount', 'nameOrig', 'oldbalanceOrg', 'newbalanceOrig',
       'nameDest', 'oldbalanceDest', 'newbalanceDest', 'isFraud']]
y = data['isFlaggedFraud']
```

```python
#Split data 70% Train 10% Validate 20% Test
from sklearn.model_selection import train_test_split
# Split data: 70% train y 30% (validate & test)
X_train, X_combined, y_train, y_combined = train_test_split(X, y, test_size=0.3, random_state=42)
print("Row count train is:", X_train.shape[0])
print("Row count validate & test is:", X_combined.shape[0])
# Split data: 30%  10% validate & 20% test
X_valid, X_test, y_valid, y_test = train_test_split(X_combined, y_combined, test_size=0.67, random_state=42)
print("Row count validate is:", X_valid.shape[0])
print("Row count test is:", X_test.shape[0])
```

```python
#subset X_valid
subset_X_valid = X_valid.iloc[:15001]
archivo_subset_valid= './data_transactions_subset_valid'
subset_X_valid.to_csv(archivo_subset_valid,index=False)
```

```python
#Save data file - format csv
archivo_train = './data_transactions_train'
X_train.to_csv(archivo_train,index=False)
archivo_valid= './data_transactions_valid'
X_valid.to_csv(archivo_valid,index=False)
archivo_test = './data_transactions_test'
X_test.to_csv(archivo_test,index=False)
```
